```python
import duckdb
import pandas as pd
```

```python
sql_query = '''
SELECT schema_name
FROM information_schema.schemata
ORDER BY schema_name;
'''

with duckdb.connect('../posts_data.duckdb') as con:
    display(con.sql(sql_query).df())
```

```python
sql_query = '''
use ingest_api_data;
show tables;
desc posts;
select user_id, id, title, char_length(title) AS title_size, body, char_length(body) AS body_size from posts where user_id > 1 limit 10;
'''

with duckdb.connect('../posts_data.duckdb') as con:
    display(con.sql(sql_query).df())
```
